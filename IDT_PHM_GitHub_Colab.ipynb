Saved-result reconstruction, distinct from prospective validation. This notebook has not been executed in a live Colab session.


```python
from pathlib import Path
import urllib.request, subprocess, sys
url='https://raw.githubusercontent.com/msalih71/idt-phm-memory-pathways/main/download_assets.py'
Path('download_assets.py').write_bytes(urllib.request.urlopen(url).read())
subprocess.run([sys.executable,'download_assets.py','--asset','paper','--extract'],check=True)
```

```python
import importlib.util
missing=[x for x in ['numpy','pandas','scipy','matplotlib'] if importlib.util.find_spec(x) is None]
if missing:subprocess.check_call([sys.executable,'-m','pip','install',*missing])
for script in ['verify_saved_results.py','analysis/build_tables.py','analysis/equilibrium_sensitivity.py','analysis/rebuild_journal_figures.py']:
    subprocess.run([sys.executable,str(Path('paper_reproduction')/script)],check=True)
```
